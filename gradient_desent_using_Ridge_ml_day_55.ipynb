```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
```

```python
X,y=load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=4)
```

```python
from sklearn.linear_model import SGDRegressor
```

```python
sgd=SGDRegressor(penalty='l2',max_iter=500,learning_rate="constant",eta0=0.1,alpha=0.001)
```

```python
sgd.fit(X_train,y_train)
y_pred=sgd.predict(X_test)
print('r2_score',r2_score(y_test,y_pred))
print(sgd.intercept_)
print(sgd.coef_)
```

```python
from sklearn.linear_model import Ridge
```

```python
reg=Ridge(alpha=0.001,max_iter=500,solver='sparse_cg')
```

```python
reg.fit(X_train,y_train)
y_pred=reg.predict(X_test)
print('r2_score',r2_score(y_test,y_pred))
print(reg.intercept_)
print(reg.coef_)
```

```python
class meraRegressor:

  def __init__(self,learning_rate,epochs,alpha):

    self.learning_rate=learning_rate
    self.epochs=epochs
    self.alpha=alpha
    self.intercept_=None
    self.coef_=None
  def fit(self,X_train,y_train):

    self.coef_=np.ones(X_train.shape[1])
    self.intercept_ =0
    theta=np.insert(self.coef_,0,self.intercept_)
    X_train=np.insert(X_train,0,1,axis=1)

    for i in range(self.epochs):
      theta_der=np.dot(X_train.T,X_train).dot(theta)-np.dot(X_train.T,y_train)+self.alpha*theta
      theta=theta-self.learning_rate*theta_der
      self.intercept_=theta[0]
      self.coef_=theta[1: ]




  def predict(self,X_test):
    return np.dot(X_test,self.coef_)+ self.intercept_
```

```python
reg1=meraRegressor(epochs=500,learning_rate=0.005,alpha=0.001)
```

```python
reg1.fit(X_train,y_train)
```

```python
y_pred=reg1.predict(X_test)
```

```python
print('r2_score',r2_score(y_test,y_pred))
print(reg1.intercept_)
print(reg1.coef_)
```
