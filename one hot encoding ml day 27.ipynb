<h1></b>one hot encoding using pandas<b></h1>

```python
pd.get_dummies(df, columns=['fuel', 'owner'],dtype=int)
```

<h1></b>k-1 one hot encoding<b></h1>

```python
pd.get_dummies(df, columns=['fuel', 'owner'],dtype=float,drop_first=True)
```

<h1></b>one hot encoding using sklearn<b></h1>

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,0:4],df.iloc[:,-1],test_size=0.2,random_state=2)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder()

X_train_new = ohe.fit_transform(
    X_train[['fuel', 'owner']]
).toarray()

X_test_new = ohe.transform(
    X_test[['fuel', 'owner']]
).toarray()

X_train_new
```

```python
np.hstack((X_train[['brand','km_driven']].values,X_train_new)).shape
```

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

ohe = OneHotEncoder(
    drop='first',
    sparse_output=False,
    dtype=np.int32
)

X_train_new = ohe.fit_transform(
    X_train[['fuel', 'owner']]
)

X_test_new = ohe.transform(
    X_test[['fuel', 'owner']]
)

X_train_new
```
