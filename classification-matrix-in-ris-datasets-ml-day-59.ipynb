```python
df=pd.read_csv('/kaggle/input/datasets/saurabh00007/iriscsv/Iris.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
df['Species']=encoder.fit_transform(df['Species'])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,0:-1],df.iloc[:,-1],test_size=0.2,random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
clf1=LogisticRegression()
clf2=DecisionTreeClassifier()
```

```python
clf1.fit(X_train,y_train)
clf2.fit(X_train,y_train)
```

```python
y_pred1=clf1.predict(X_test)
y_pred2=clf2.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
print('accuracy_score in logisticRegression',accuracy_score(y_test,y_pred1))
print('accuracy_score in decision_tree',accuracy_score(y_test,y_pred2))
```

```python
print('confusion matrix in logisticRegression /n')
pd.DataFrame(confusion_matrix(y_test,y_pred1),columns=list(range(0,3)))
```

```python
print('confusion matrix in DecisiontreeClassifier /n')
pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,3)))
```

```python
result=pd.DataFrame()
result['Actual Result']=y_test
result['LogisticRegression prediction']=y_pred1
result['decisiomnTreeClassifier prerdiction']= y_pred2
result.sample(10)
```
