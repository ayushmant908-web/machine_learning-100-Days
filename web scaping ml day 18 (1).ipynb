```python
salary = []
interview = []
jobs = []

cards = soup.find_all("div", class_="companyCardWrapper")

for card in cards:
    counts = card.find_all("span", class_="companyCardWrapper__ActionCount")

    if len(counts) >= 4:
        salary.append(counts[1].text.strip())
        interview.append(counts[2].text.strip())
        jobs.append(counts[3].text.strip())

print("Salary:", salary)
print("Interview:", interview)
print("Jobs:", jobs)
```

```python
final = pd.DataFrame()

headers = {
    "User-Agent": "Mozilla/5.0"
}

for j in range(1, 101):

    url = f"https://www.ambitionbox.com/list-of-companies?page={j}"

    webpage = requests.get(url, headers=headers).text

    soup = BeautifulSoup(webpage, "lxml")

    cards = soup.find_all("div", class_="companyCardWrapper")

    company = []
    rating = []
    reviews = []
    company_type = []
    head_office = []
    salaries = []
    interviews = []
    jobs = []

    for card in cards:

        # Company Name
        try:
            company.append(card.find("h2").text.strip())
        except:
            company.append(np.nan)

        # Rating
        try:
            rating.append(card.find("div", class_="rating_text").text.strip())
        except:
            rating.append(np.nan)

        # Reviews
        try:
            reviews.append(card.find("span", class_="companyCardWrapper__companyRatingCount").text.strip())
        except:
            reviews.append(np.nan)

        # Company Type & Head Office
        try:
            info = card.find("span", class_="companyCardWrapper__interLinking").text.strip()
            company_type.append(info.split("|")[0].strip())
            head_office.append(info.split("|")[1].strip())
        except:
            company_type.append(np.nan)
            head_office.append(np.nan)

        # Salaries, Interviews, Jobs
        try:
            counts = card.find_all("span", class_="companyCardWrapper__ActionCount")
            salaries.append(counts[1].text.strip())
            interviews.append(counts[2].text.strip())
            jobs.append(counts[3].text.strip())
        except:
            salaries.append(np.nan)
            interviews.append(np.nan)
            jobs.append(np.nan)

    df = pd.DataFrame({
        "Company": company,
        "Rating": rating,
        "Reviews": reviews,
        "Company_Type": company_type,
        "Head_Office": head_office,
        "Salaries": salaries,
        "Interviews": interviews,
        "Jobs": jobs
    })

    final = pd.concat([final, df], ignore_index=True)

print(final.shape)
```

```python
final = pd.DataFrame()

for page in range(1, 101):

    url = f"https://www.ambitionbox.com/list-of-companies?page={page}"
    webpage = requests.get(url, headers=headers).text

    soup = BeautifulSoup(webpage, "lxml")

    cards = soup.find_all("div", class_="companyCardWrapper")

    company = []
    reviews = []
    company_type = []
    head_office = []
    salaries = []
    interviews = []
    jobs = []

    for card in cards:

        # Company Name
        try:
            company.append(card.find("h2").text.strip())
        except:
            company.append(np.nan)

        # Rating
        

        # Reviews
        try:
            reviews.append(card.find("span", class_="companyCardWrapper__companyRatingCount").text.strip())
        except:
            reviews.append(np.nan)

        # Company Type & Head Office
        try:
            info = card.find("span", class_="companyCardWrapper__interLinking").text.strip()
            parts = [x.strip() for x in info.split("|")]

            company_type.append(parts[0] if len(parts) > 0 else np.nan)
            head_office.append(parts[1] if len(parts) > 1 else np.nan)

        except:
            company_type.append(np.nan)
            head_office.append(np.nan)

        # Salaries, Interviews, Jobs
        try:
            counts = card.find_all("span", class_="companyCardWrapper__ActionCount")

            salaries.append(counts[1].text.strip() if len(counts) > 1 else np.nan)
            interviews.append(counts[2].text.strip() if len(counts) > 2 else np.nan)
            jobs.append(counts[3].text.strip() if len(counts) > 3 else np.nan)

        except:
            salaries.append(np.nan)
            interviews.append(np.nan)
            jobs.append(np.nan)

    df = pd.DataFrame({
        "Company": company,
        
        "Reviews": reviews,
        "Company_Type": company_type,
        "Head_Office": head_office,
        "Salaries": salaries,
        "Interviews": interviews,
        "Jobs": jobs
    })

    final = pd.concat([final, df], ignore_index=True)

print(final.shape)
final.head()
```
