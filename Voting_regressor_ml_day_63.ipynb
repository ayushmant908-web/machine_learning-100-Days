```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import fetch_openml
boston = fetch_openml(
    name='boston',
    version=1,
    as_frame=True
)
```

```python
df=boston.frame
```

```python
df.shape
```

```python
X=df.iloc[:,0:13]
y=df.iloc[:,-1]
```

```python
X.shape
```

```python
y.shape
```

```python
X
```

```python
X = pd.get_dummies(X, drop_first=True, dtype=int)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import cross_val_score
```

```python
lr=LinearRegression()
dt=DecisionTreeRegressor()
svr=SVR()
```

```python
estimators=[('lr',lr),('dt',dt),('svr',svr)]
```

```python
for estimator in estimators:
  x=cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(x),2))
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
X = pd.get_dummies(X, drop_first=True, dtype=int)
```

```python
vr=VotingRegressor(estimators)
scorers=cross_val_score(vr,X,y,cv=10,scoring='r2')
print('voting_regrtessor',np.round(np.mean(scorers),2))
```

```python
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
```

```python
# using the same algorithm
dv1=DecisionTreeRegressor(max_depth=1)
dv2=DecisionTreeRegressor(max_depth=2)
dv3=DecisionTreeRegressor(max_depth=3)
dv4=DecisionTreeRegressor(max_depth=4)
dv5=DecisionTreeRegressor(max_depth=None)
```

```python
estimators=[('dv1',dv1),('dv2',dv2),('dv3',dv3),('dv4',dv4),('dv5',dv5)]
```

```python

for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
```

```python
vc=VotingRegressor(estimators)
scores=cross_val_score(vc,X,y,scoring='r2',cv=10)
print('voting_regressor',np.round(np.mean(scores),2))
```
