```python
df=pd.read_csv('/kaggle/input/competitions/digit-recognizer/train.csv')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
clf1=LogisticRegression()
clf2=DecisionTreeClassifier()
```

```python
clf1.fit(X_train,y_train)
clf2.fit(X_train,y_train)
```

```python
y_pred1=clf1.predict(X_test)
y_pred2=clf2.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print('logisticRegression accuracy_score',accuracy_score(y_test,y_pred1))
print('decision Tree classifier',accuracy_score(y_test,y_pred2))
```

```python
print(' logisticRegression confusion_matrix \n')
pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,10)))
```

```python
print('Desicion Tree Classifier confusion matrix/n')
pd.DataFrame(confusion_matrix(y_test,y_pred2),columns=list(range(0,10)))
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score
```

<h1><b>Average have two hidden pairamite(avg=weighted,None,macro)

```python
precision_score(y_test,y_pred1,average='weighted')
```

```python
recall_score(y_test,y_pred1,average='weighted')
```

```python
f1_score(y_test,y_pred1,average='weighted')
```

```python
from sklearn.metrics import classification_report
print('classsification_report matrix',classification_report(y_test,y_pred1))
```
