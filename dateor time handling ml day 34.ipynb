<h1></b>Working with date <b></h1>

```python
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

<h2>Extract the year</h2>

```python
date['date_year']=date['date'].dt.year
```

```python
date.sample(5)
```

<h2>Extract the month</h2>

```python
date['date_month_name'] = date['date'].dt.month_name()

date.head()
```

<h2>Extract the day_num</h2>

```python
date['date_day']=date['date'].dt.day
```

```python
date.sample(5)
```

<h2>Extract the day</h2>

```python
date['date_day']=date['date'].dt.day_name()
```

```python
date.sample(5)
```

<h2>Extract the day_of_week</h2>

```python
date['date_of now']=date['date'].dt.dayofweek
date.head()
```

```python
date['week day name']=date['date'].dt.day_name()
date.head()
```

```python
date=date.drop(columns=['product_id','city_id','orders'])
```

```python
date.head()
```

<h2>Extract the _weekend</h2>

```python
date['wekend']=np.where(date['date_day'].isin(['Sunday','Saturday']),1,0)
```

```python
date.sample(5)
```

<h2>Extract the _week_num</h2>

```python
date['date_week'] = date['date'].dt.isocalendar().week

date.head()
```

<h2>Extract the quarter</h2>

```python
date['quarter'] = date['date'].dt.quarter
date.head()
```

<h2>Extract the semster</h2>

```python
date['semester']=np.where(date['quarter'].isin([1,2]),1,2)
date.head()
```

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python

(today - date['date']).dt.days
```

```python
# Months passed

np.round((today -date['date']) / np.timedelta64(1, 'm'),0)
```
