```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt
```

```python
X,y=make_regression(n_samples=100,n_features=1,n_informative=1,n_targets=1,noise=20,random_state=13)
```

```python
plt.scatter(X,y)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(X,y)
print('coef_',lr.coef_)
print('intercept_',lr.intercept_)
```

```python
from sklearn.linear_model import Lasso
```

```python
lass=Lasso(alpha=1)
```

```python
lass.fit(X,y)
print('coef_',lass.coef_)
print('intercept_',lass.intercept_)
```

```python
lass1=Lasso(alpha=10)
```

```python
lass1.fit(X,y)
print('coef_',lass1.coef_)
print('intercept_',lass1.intercept_)
```

```python
lass2=Lasso(alpha=15)
```

```python
lass2.fit(X,y)
print('coef',lass2.coef_)
print('interfcept_',lass2.intercept_)
```

```python
plt.plot(X,y,'b. ')
plt.plot(X,lass.predict(X),color='red',label='alpha=1')
plt.plot(X,lass1.predict(X),color='green',label='alpha=10')
plt.plot(X,lass2.predict(X),color='blue',label='alpha=15')
plt.legend()
plt.show()
```
